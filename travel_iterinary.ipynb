```python
# ======================================================
# Travel Itinerary Planner
# Author: Richa Dhiman
#
# What this program does:
# - Welcomes the user nicely
# - Creates a 7-day itinerary for popular places
# - Calculates realistic travel budget
# - Supports hotel/hostel + food styles
# - Shows step-by-step budget calculation if user asks
# - Saves plan + budget to a JSON file
# ======================================================

import json


# ------------------------------------------------------
# BASIC DESTINATION DATA
# ------------------------------------------------------
places_data = {
    "Manali": "Hill station with snow & adventure",
    "Goa": "Beaches, nightlife & relaxation",
    "Jaipur": "Forts, palaces & culture",
    "Munnar": "Tea gardens & greenery",
    "Varanasi": "Spiritual city on the Ganges"
}


# ------------------------------------------------------
# STAY PRICES (PER NIGHT)
# ------------------------------------------------------
price_table = {
    "Manali": {"hotel": 2000, "hostel": 600},
    "Goa": {"hotel": 2200, "hostel": 500},
    "Jaipur": {"hotel": 1700, "hostel": 450},
    "Munnar": {"hotel": 1600, "hostel": 400},
    "Varanasi": {"hotel": 1400, "hostel": 350}
}


# ------------------------------------------------------
# FOOD STYLES (PER DAY)
# ------------------------------------------------------
food_styles = {
    "street": 200,
    "cafe": 450,
    "premium": 900
}


# ------------------------------------------------------
# 7-DAY ATTRACTIONS FOR EACH PLACE
# ------------------------------------------------------
attractions_short = {

    "Manali": [
        {"name": "Hadimba Temple", "type": "spiritual", "cost": 0, "tips": "Peaceful forest area"},
        {"name": "Old Manali Cafe Walk", "type": "relax", "cost": 500, "tips": "Try local cafés"},
        {"name": "Solang Valley", "type": "adventure", "cost": 1200, "tips": "Paragliding & snow"},
        {"name": "Atal Tunnel & Sissu", "type": "nature", "cost": 1500, "tips": "Scenic drive"},
        {"name": "Jogini Waterfall Trek", "type": "nature", "cost": 0, "tips": "Wear trekking shoes"},
        {"name": "Vashisht Hot Springs", "type": "relax", "cost": 0, "tips": "Natural hot water"},
        {"name": "Mall Road Shopping", "type": "shopping", "cost": 800, "tips": "Buy souvenirs"}
    ],

    "Goa": [
        {"name": "Baga & Calangute Beach", "type": "relax", "cost": 0, "tips": "Evenings lively"},
        {"name": "Anjuna Flea Market", "type": "shopping", "cost": 300, "tips": "Bargain well"},
        {"name": "Dudhsagar Waterfall", "type": "nature", "cost": 1200, "tips": "Full day trip"},
        {"name": "Water Sports", "type": "adventure", "cost": 1500, "tips": "Licensed vendors"},
        {"name": "Old Goa Churches", "type": "history", "cost": 0, "tips": "UNESCO sites"},
        {"name": "Chapora Fort Sunset", "type": "nature", "cost": 0, "tips": "Amazing views"},
        {"name": "Beachside Dinner", "type": "food", "cost": 1000, "tips": "Seafood recommended"}
    ],

    "Jaipur": [
        {"name": "Amber Fort", "type": "history", "cost": 500, "tips": "Go early"},
        {"name": "City Palace", "type": "history", "cost": 300, "tips": "Museum inside"},
        {"name": "Hawa Mahal", "type": "history", "cost": 200, "tips": "Best photos in morning"},
        {"name": "Jantar Mantar", "type": "history", "cost": 200, "tips": "Astronomy site"},
        {"name": "Albert Hall Museum", "type": "history", "cost": 200, "tips": "Evening lights"},
        {"name": "Bapu Bazaar Shopping", "type": "shopping", "cost": 800, "tips": "Local handicrafts"},
        {"name": "Chokhi Dhani Cultural Night", "type": "food", "cost": 1200, "tips": "Rajasthani thali"}
    ],

    "Munnar": [
        {"name": "Tea Garden Walk", "type": "nature", "cost": 0, "tips": "Morning best"},
        {"name": "Eravikulam National Park", "type": "nature", "cost": 300, "tips": "Nilgiri Tahr"},
        {"name": "Mattupetty Dam", "type": "nature", "cost": 200, "tips": "Boating"},
        {"name": "Echo Point", "type": "nature", "cost": 0, "tips": "Fun echoes"},
        {"name": "Attukad Waterfalls", "type": "nature", "cost": 0, "tips": "Monsoon best"},
        {"name": "Spice Plantation Tour", "type": "food", "cost": 500, "tips": "Guided tour"},
        {"name": "Local Market & Café", "type": "relax", "cost": 400, "tips": "Souvenirs"}
    ],

    "Varanasi": [
        {"name": "Ganga Aarti", "type": "spiritual", "cost": 0, "tips": "Reach early"},
        {"name": "Sunrise Boat Ride", "type": "nature", "cost": 300, "tips": "Peaceful"},
        {"name": "Kashi Vishwanath Temple", "type": "spiritual", "cost": 0, "tips": "Dress modestly"},
        {"name": "Sarnath", "type": "history", "cost": 200, "tips": "Buddhist site"},
        {"name": "Assi Ghat Walk", "type": "relax", "cost": 0, "tips": "Calm vibes"},
        {"name": "Street Food Walk", "type": "food", "cost": 300, "tips": "Try kachori"},
        {"name": "Silk Shopping", "type": "shopping", "cost": 1000, "tips": "Banarasi sarees"}
    ]
}


# ------------------------------------------------------
# CREATE ITINERARY
# ------------------------------------------------------
def create_itinerary(place, days):
    activities = attractions_short[place]
    plan = {}

    for d in range(days):
        plan[f"Day {d+1}"] = [activities[d]]

    return plan


# ------------------------------------------------------
# BUDGET CALCULATOR (TRANSPARENT)
# ------------------------------------------------------
def calculate_budget(place, days, stay_type, food_choice, itinerary):
    stay_price = price_table[place][stay_type]
    stay_cost = stay_price * days

    food_per_day = food_styles[food_choice]
    food_cost = food_per_day * days

    activity_cost = 0
    for acts in itinerary.values():
        for a in acts:
            activity_cost += a["cost"]

    local_travel_per_day = 300
    local_travel_cost = local_travel_per_day * days

    total = stay_cost + food_cost + activity_cost + local_travel_cost

    return {
        "stay_price": stay_price,
        "stay_cost": stay_cost,
        "food_per_day": food_per_day,
        "food_cost": food_cost,
        "activity_cost": activity_cost,
        "local_travel_per_day": local_travel_per_day,
        "local_travel_cost": local_travel_cost,
        "total_budget": total
    }


# ------------------------------------------------------
# SAVE TO JSON
# ------------------------------------------------------
def save_plan(place, days, itinerary, budget):
    filename = f"{place.lower()}_{days}days_plan.json"
    with open(filename, "w") as f:
        json.dump({
            "place": place,
            "days": days,
            "itinerary": itinerary,
            "budget": budget
        }, f, indent=2)

    print("\n📁 Plan saved as:", filename)


# ------------------------------------------------------
# MAIN PROGRAM
# ------------------------------------------------------
print("\n🌍✨ Welcome to the Travel Itinerary Planner ✨🌍")
print("Let’s plan a beautiful trip for you!\n")

print("📍 Available destinations:")
for p in places_data:
    print(" •", p)

place = input("\nWhere would you like to go? ").strip().title()
days = int(input("How many days will you stay? (max 7): "))

stay_type = input("Stay type (hotel / hostel): ").strip().lower()
food_choice = input("Food style (street / cafe / premium): ").strip().lower()

# Create itinerary
itinerary = create_itinerary(place, days)

# Calculate budget
budget = calculate_budget(place, days, stay_type, food_choice, itinerary)


# ------------------------------------------------------
# OUTPUT
# ------------------------------------------------------
print(f"\n✨ {days}-Day Itinerary for {place} ✨\n")

for day, acts in itinerary.items():
    print(day + ":")
    for a in acts:
        print(f"  - {a['name']} ({a['type']})")
        print(f"    Tip: {a['tips']}")
    print()

print("💰 --- Budget Summary ---")
print(f"🏨 Stay ({stay_type.title()}): ₹ {budget['stay_cost']}")
print(f"🍽️ Food ({food_choice.title()}): ₹ {budget['food_cost']}")
print(f"🎟️ Activities: ₹ {budget['activity_cost']}")
print(f"🚕 Local Travel: ₹ {budget['local_travel_cost']}")
print("--------------------------------")
print(f"💵 Total Budget: ₹ {budget['total_budget']}")

# Ask if user wants calculation steps
show_calc = input("\nDo you want to see full calculation? (y/n): ").strip().lower()

if show_calc == "y":
    print("\n🧮 --- Detailed Calculation ---")
    print(f"Stay: {budget['stay_price']} × {days} = ₹ {budget['stay_cost']}")
    print(f"Food: {budget['food_per_day']} × {days} = ₹ {budget['food_cost']}")
    print(f"Local travel: {budget['local_travel_per_day']} × {days} = ₹ {budget['local_travel_cost']}")
    print(f"Activities total = ₹ {budget['activity_cost']}")
    print("--------------------------------")
    print(f"FINAL TOTAL = ₹ {budget['total_budget']}")

save_plan(place, days, itinerary, budget)
```
